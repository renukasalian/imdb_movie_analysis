```python
# Name the director who has made the most movies 

cursor.execute("""
                    SELECT d.name, COUNT(m.id) AS movie_count
                    FROM movies AS m
                    JOIN directors AS d
                        ON m.director_id = d.id
                    GROUP BY m.director_id, d.name
                    ORDER BY movie_count DESC
                    LIMIT 1;
                """)

most_movie = cursor.fetchall()
print(f"The director {most_movie[0][0]} has made {most_movie[0][1]} movies")
```

```python
# Name of the director who is most bankable

cursor.execute(""" 
               SELECT d.name 
               FROM directors as d 
               JOIN movies as m
                    ON m.director_id = d.id
               GROUP BY d.id
               ORDER BY SUM(m.budget) DESC
               LIMIT 1 ;
               """)

most_bankable_director = cursor.fetchall()
print(f"Most bankable director is {most_bankable_director[0][0]}")
```

visualization


```python
# Budget vs Revenue-shows whether higher budgets lead to higher revenue.
plt.figure(figsize=(8,5))
sns.scatterplot(data=movie_data, x='budget', y='revenue')
plt.title('Budget vs Revenue')
plt.xlabel('Budget')
plt.ylabel('Revenue')
plt.tight_layout()
plt.show()
```

```python
# Top 10 revenue generating movies
top10_revenue = movie_data.nlargest(10, 'revenue')

plt.figure(figsize=(10,5))
sns.barplot(data=top10_revenue, x='title', y='revenue')
plt.xticks(rotation=70)
plt.title('Top 10 Revenue-Generating Movies')
plt.xlabel('Movie Title')
plt.ylabel('Revenue')
plt.tight_layout()
plt.show()
```

```python
# Top 10 Budget Movies
top10_budget = movie_data.nlargest(10, 'budget')

plt.figure(figsize=(10,5))
sns.barplot(data=top10_budget, x='title', y='budget')
plt.xticks(rotation=80)
plt.title('Top 10 Highest-Budget Movies')
plt.xlabel('Movie Title')
plt.ylabel('Budget')
plt.tight_layout()
plt.show()
```
